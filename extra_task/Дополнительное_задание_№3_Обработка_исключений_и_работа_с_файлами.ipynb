### **Пункт №5. Поиск слова в файле**

Напишите программу, которая открывает файл `output.txt` и проверяет, содержится ли в нём слово, введённое пользователем.

**Обязательно:**

- создайте файл минимум с 10 строками текста;
- выведите содержимое файла;
- проверьте поиск минимум 5 разных слов;
- выведите номера строк, в которых найдено слово; номера получайте функцией `enumerate()`;
- выведите количество вхождений слова.

> 📚 **Подсказка:** `for number, line in enumerate(f, start=1)` нумерует строки файла с единицы — см. раздел №12, «Номера строк: функция `enumerate()`».

````python
with open('output.txt', 'w+', encoding='utf-8') as f:
  for i in 'abcdefglvxs':
    f.write(f'{i}\n')
  f.seek(0)
  print(f.read())

def search_word(filename, word):

    lines_with_word = []
    count = 0

    with open(filename, 'r', encoding='utf-8') as f:
        for number, line in enumerate(f, start=1):
            count_in_line = line.count(word)

            if count_in_line > 0:
                lines_with_word.append(number)
                count += count_in_line
                print(f"Строка {number}: найдено {count_in_line} раз")

    if lines_with_word:
        print(f"Найдено в строках: {lines_with_word}")
        print(f"Всего вхождений: {count}")
    else:
        print(f"Слово '{word}' не найдено в файле")

words_to_search = ['a', 'dd', 'b', 'e', '1']

for word in words_to_search:
    search_word('output.txt', word)
````

- обработайте случай несуществующего файла.

Текст для тестового файла:

```python
test_content = """Первая строка текста
Вторая строка с большим количеством слов
Третья строка
Четвертая строка для теста
Пятая и последняя строка"""
```

````python
test_content = """Первая строка текста
Вторая строка с большим количеством слов
Третья строка
Четвертая строка для теста
Пятая и последняя строка"""

with open('test.txt', 'w+', encoding='utf-8') as f:
  f.write(test_content)

def file_statistics(filename):
  try:
    with open(f'{filename}', 'r', encoding='utf-8') as f:
      print(f.read())
      f.seek(0)
      c_ = 0
      c = 0
      word = []
      for number, line in enumerate(f, start=1):
        for i in line:
          if i == ' ' or i == '\n':
            c_ += 1
          c += 1

        line = line.strip()
        word += line.split()

      print(f'Строки: {number}\nС пробелом: {c}\nБез пробела: {c-c_}\nКол-во слов: {len(word)}')
  except FileNotFoundError:
    print('Файла нету')

file_statistics('test.txt')


````

### **Пункт №8. Поиск и замена в файле**

Напишите функцию `replace_in_file(filename, old_text, new_text)`, которая заменяет все вхождения `old_text` на `new_text` и сохраняет результат в новый файл с префиксом `modified_` (например, `modified_output.txt`).

**Обязательно:**

- создайте файл с текстом, в котором заменяемое слово встречается минимум 3 раза;
- выведите содержимое файла до замены;
- выполните замену;
- выведите содержимое нового файла после замены;
- выведите количество произведённых замен.

````python
def replace_in_file(filename, old_text, new_text):
    with open(filename, 'r', encoding='utf-8') as f:
        text = f.read()
    count = text.count(old_text)
    new_content = text.replace(old_text, new_text)
    new_filename = 'modified_' + filename
    with open(new_filename, 'w', encoding='utf-8') as f:
        f.write(new_content)
    return count

text = 'яблоко, груша, яблоко, слива, яблоко'
with open('output.txt', 'w', encoding='utf-8') as f:
    f.write(text)

with open('output.txt', 'r', encoding='utf-8') as f:
    print('До замены:')
    print(f.read())

count = replace_in_file('output.txt', 'яблоко', 'апельсин')
print('Количество замен:', count)

with open('modified_output.txt', 'r', encoding='utf-8') as f:
    print('После замены:')
    print(f.read())
````

### **Пункт №9. Объединение файлов**

Создайте функцию `merge_files(file_list, output_file)`, которая объединяет содержимое нескольких текстовых файлов в один и добавляет после содержимого каждого файла разделитель.

**Обязательно:**

- создайте минимум 3 файла с разным содержимым;
- выведите содержимое каждого исходного файла;
- выполните объединение;
- выведите содержимое итогового файла.

````python
def merge_files(file_list, output_file):
    with open(output_file, 'w', encoding='utf-8') as out:
        for file in file_list:
            with open(file, 'r', encoding='utf-8') as f:
                out.write(f.read())
                out.write('\n---\n')

with open('f1.txt', 'w', encoding='utf-8') as f:
    f.write('первый файл\n')
with open('f2.txt', 'w', encoding='utf-8') as f:
    f.write('второй файл\n')
with open('f3.txt', 'w', encoding='utf-8') as f:
    f.write('третий файл\n')

print('f1:', open('f1.txt', encoding='utf-8').read())
print('f2:', open('f2.txt', encoding='utf-8').read())
print('f3:', open('f3.txt', encoding='utf-8').read())

merge_files(['f1.txt', 'f2.txt', 'f3.txt'], 'merged.txt')

with open('merged.txt', 'r', encoding='utf-8') as f:
    print('Итоговый файл:')
    print(f.read())
````

````python
# @title Результаты автоматической проверки
# AUTOCHECKER_A3
import contextlib as _contextlib
import html as _html
import io as _io
import json as _json
import sys as _sys
import types as _types
import urllib.request as _urllib

from IPython.display import HTML as _HTML, display as _display

_TASK_IDS = ['a3_01',
 'a3_02',
 'a3_03',
 'a3_04',
 'a3_05',
 'a3_06',
 'a3_07',
 'a3_08',
 'a3_09',
 'a3_10',
 'a3_11',
 'a3_12']
_TASK_SOURCE_IDS = {'a3_01': ['724c5954'],
 'a3_02': ['31448e09'],
 'a3_03': ['1c067cf3'],
 'a3_04': ['cbce6c5f'],
 'a3_05': ['df27a986'],
 'a3_06': ['84aa205a'],
 'a3_07': ['31bd8a06'],
 'a3_08': ['d961bbf3'],
 'a3_09': ['0c262bd2'],
 'a3_10': ['43bfd6b8'],
 'a3_11': ['64a05e35'],
 'a3_12': ['581b5868']}
_TASK_EDITABLE_IDS = {'a3_01': ['724c5954'],
 'a3_02': ['31448e09'],
 'a3_03': ['1c067cf3'],
 'a3_04': ['cbce6c5f'],
 'a3_05': ['df27a986'],
 'a3_06': ['84aa205a'],
 'a3_07': ['31bd8a06'],
 'a3_08': ['d961bbf3'],
 'a3_09': ['0c262bd2'],
 'a3_10': ['43bfd6b8'],
 'a3_11': ['64a05e35'],
 'a3_12': ['581b5868']}
_TASK_SOURCE_POSITIONS = {'a3_01': [0],
 'a3_02': [1],
 'a3_03': [2],
 'a3_04': [3],
 'a3_05': [4],
 'a3_06': [5],
 'a3_07': [6],
 'a3_08': [7],
 'a3_09': [8],
 'a3_10': [9],
 'a3_11': [10],
 'a3_12': [11]}
_TASK_EDITABLE_POSITIONS = {'a3_01': [0],
 'a3_02': [1],
 'a3_03': [2],
 'a3_04': [3],
 'a3_05': [4],
 'a3_06': [5],
 'a3_07': [6],
 'a3_08': [7],
 'a3_09': [8],
 'a3_10': [9],
 'a3_11': [10],
 'a3_12': [11]}
_CHECKER_CELL_ID = "autoCheckA3"
_CHECKER_URL = "https://raw.githubusercontent.com/tester170/python_programming_autotest/main/answer_checker.py"


def _notebook_from_reply(value):
    if isinstance(value, str):
        value = _json.loads(value)
    if not isinstance(value, dict):
        raise TypeError("Colab вернул данные блокнота в неизвестном формате")
    if isinstance(value.get("cells"), list):
        return value
    for key in ("ipynb", "notebook", "data"):
        nested = value.get(key)
        if isinstance(nested, str):
            try:
                nested = _json.loads(nested)
            except _json.JSONDecodeError:
                continue
        if isinstance(nested, dict) and isinstance(nested.get("cells"), list):
            return nested
    raise ValueError("Не удалось получить ячейки открытого блокнота")


def _cell_source(cell):
    source = cell.get("source", "")
    return "".join(source) if isinstance(source, list) else str(source)


def _cell_id(cell):
    return cell.get("id") or cell.get("metadata", {}).get("id")


def _solution_is_empty(code):
    return not any(
        line.strip() and not line.lstrip().startswith("#")
        for line in code.splitlines()
    )


def _status(result):
    if result is None:
        return "⚠️ Решение не заполнено"
    if result["ok"]:
        return f"✅ Верно ({result['passed']}/{result['total']})"
    if result["error"] == "timeout":
        return f"⏱️ Превышено время выполнения ({result['passed']}/{result['total']})"
    if result["error"] == "runtime":
        return f"❌ Ошибка выполнения ({result['passed']}/{result['total']})"
    return f"❌ Есть ошибка ({result['passed']}/{result['total']})"


try:
    from google.colab import _message as _colab_message

    _reply = _colab_message.blocking_request("get_ipynb", request="", timeout_sec=30)
    _notebook = _notebook_from_reply(_reply)

    _request = _urllib.Request(
        _CHECKER_URL,
        headers={"User-Agent": "python-programming-autochecker"},
    )
    with _urllib.urlopen(_request, timeout=20) as _response:
        _checker_source = _response.read().decode("utf-8")

    _checker_name = "_python_programming_autochecker"
    _checker = _types.ModuleType(_checker_name)
    _checker.__file__ = _CHECKER_URL
    _sys.modules[_checker_name] = _checker
    with _contextlib.redirect_stdout(_io.StringIO()), _contextlib.redirect_stderr(_io.StringIO()):
        exec(compile(_checker_source, _CHECKER_URL, "exec"), _checker.__dict__)

    if getattr(_checker, "CHECKER_VERSION", 0) < 3:
        raise RuntimeError("Сначала замените содержимое репозитория автопроверки файлами из подготовленного архива")
    _missing_tasks = [task_id for task_id in _TASK_IDS if task_id not in _checker.TASKS]
    if _missing_tasks:
        raise RuntimeError("В репозитории автопроверки отсутствуют тесты для этого блокнота")

    _code_cells = []
    _code_by_id = {}
    for _cell in _notebook.get("cells", []):
        if _cell.get("cell_type") != "code":
            continue
        _source = _cell_source(_cell)
        _id = _cell_id(_cell)
        if _id == _CHECKER_CELL_ID or "AUTOCHECKER_" in _source:
            continue
        _code_cells.append(_source)
        if _id:
            _code_by_id[_id] = _source

    def _resolve_sources(ids, positions):
        by_id = [_code_by_id.get(cell_id) for cell_id in ids]
        if all(source is not None for source in by_id):
            return by_id
        try:
            return [_code_cells[position] for position in positions]
        except IndexError as error:
            raise RuntimeError("Не удалось найти ячейки заданий в блокноте") from error

    _rows = []
    _passed_tasks = 0
    for _index, _task_id in enumerate(_TASK_IDS, 1):
        _sources = _resolve_sources(
            _TASK_SOURCE_IDS[_task_id], _TASK_SOURCE_POSITIONS[_task_id]
        )
        _editable = _resolve_sources(
            _TASK_EDITABLE_IDS[_task_id], _TASK_EDITABLE_POSITIONS[_task_id]
        )
        _code = "\n\n".join(_sources)
        _result = None if _solution_is_empty("\n\n".join(_editable)) else _checker.check_task(_task_id, _code)
        if _result is not None and _result["ok"]:
            _passed_tasks += 1
        _title = _checker.TASKS[_task_id].title
        _rows.append(
            "<tr>"
            f"<td style='padding:7px 10px;border-bottom:1px solid #eee'>{_index}</td>"
            f"<td style='padding:7px 10px;border-bottom:1px solid #eee'>{_html.escape(_title)}</td>"
            f"<td style='padding:7px 10px;border-bottom:1px solid #eee'>{_html.escape(_status(_result))}</td>"
            "</tr>"
        )

    _report = f"""
    <div style="font-family:Arial,sans-serif;max-width:850px">
      <h3 style="margin:0 0 8px">Результаты автоматической проверки</h3>
      <table style="border-collapse:collapse;width:100%;border:1px solid #ddd">
        <thead><tr style="background:#f5f5f5">
          <th style="padding:8px;text-align:left">№</th>
          <th style="padding:8px;text-align:left">Задание</th>
          <th style="padding:8px;text-align:left">Результат</th>
        </tr></thead>
        <tbody>{''.join(_rows)}</tbody>
      </table>
      <div style="margin-top:12px;font-size:18px"><b>Итого: {_passed_tasks} из {len(_TASK_IDS)}</b></div>
    </div>
    """
    _display(_HTML(_report))
except Exception as _error:
    _display(_HTML(
        "<div style='font-family:Arial,sans-serif;padding:12px;border:1px solid #e0a0a0;background:#fff7f7'>"
        "⚠️ Не удалось запустить автоматическую проверку: "
        + _html.escape(str(_error))
        + "</div>"
    ))
````
